```python
def area_triangulo(a, b, c):
  #Calculara el area de un triangulo con lados, a, b, c con la formula de Heron
  #Variables Locales
  s=(a+b+c)/2 #formula del semiperimetro
  area=(s*(s-a)*(s-b)*(s-c))**0.5 #formula del area de Heron
  return area
#El usuario eligira valores para a, b y c
a=float(input("Ingrese el valor del lado a: "))
b=float(input("Ingrese el valor del lado b: "))
c=float(input("Ingrese el valor del lado c: "))
#Verificando que los valores dados sean positivos
while a<=0 or b<=0 or c<=0:
  print("Los lados deben ser positivos, intentelo de nuevo")
  a=float(input("Ingrese el valor del lado a: "))
  b=float(input("Ingrese el valor del lado b: "))
  c=float(input("Ingrese el valor del lado c: "))
#Verificando que los valores cumplan la desigualdad del triangulo
while not(a+b>c and a+c>b and b+c>a):
  print("Los lados no forman un triangulo, intentalo de nuevo")
  a=float(input("Ingrese el valor del lado a: "))
  b=float(input("Ingrese el valor del lado b: "))
  c=float(input("Ingrese el valor del lado c: "))
resultado = area_triangulo(a, b, c)
print("El area del triangulo es: ", resultado)
```
