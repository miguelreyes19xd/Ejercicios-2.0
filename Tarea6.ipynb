```python
import numpy as np
import matplotlib.pyplot as plt
def escoger_matriz():
    # El usuario escogera las dimensiones
    r = int(input("Número de renglones: "))
    c = int(input("Número de columnas: "))
    print(f"Introduce los valores de la matriz ({r}x{c}), fila por fila:")
    matriz = []
    for i in range(r):
        fila = input(f"Fila {i+1} (separa los números con espacios): ").split()
        # Convertir texto a float
        fila = [float(x) for x in fila]
        # Validar longitud correcta
        if len(fila) != c:
            raise ValueError("La cantidad de elementos no coincide con las columnas.")
        matriz.append(fila)
    return np.array(matriz)
def sumar_matrices(A, B):
    if A.shape != B.shape:
        raise ValueError("Las matrices deben tener las mismas dimensiones para sumarse.")
    return A + B
#Programa principal
print("Matriz A")
A = escoger_matriz()
print("Matriz B")
B = escoger_matriz()
print("Resultado de A + B")
print(sumar_matrices(A, B))
```

```python
def esc_matriz():
    r = int(input("Número de renglones: "))
    c = int(input("Número de columnas: "))
    print(f"Introduce los {r} renglones de la matriz (cada uno con {c} números):")
    matriz = []
    for i in range(r):
        fila = input(f"Fila {i+1}: ").split()
        fila = [float(x) for x in fila]
        if len(fila) != c:
            raise ValueError("Número incorrecto de columnas.")
        matriz.append(fila)
    return np.array(matriz)
def producto_matricial(A, B):
    # Validar que columnas de A = renglones de B
    if A.shape[1] != B.shape[0]:
        raise ValueError("Las matrices no son compatibles."
                         "Se requiere: columnas de A = renglones de B.")
    return np.matmul(A, B)


#Programa principal
print("Matriz A")
A = esc_matriz()

print("Matriz B")
B = esc_matriz()

print("A @ B")
print(producto_matricial(A, B))
```

```python
def leer_matriz_cuadrada():
    n = int(input("Tamaño de la matriz (n para matriz n x n): "))
    print(f"Introduce los valores de la matriz ({n}x{n}), fila por fila:")
    matriz = []
    for i in range(n):
        fila = input(f"Fila {i+1}: ").split()
        fila = [float(x) for x in fila]
        if len(fila) != n:
            raise ValueError("La fila no tiene la cantidad correcta de elementos.")
        matriz.append(fila)
    return np.array(matriz)
def traza(A):
    return np.trace(A)
#Programa principal
print("Matriz A")
A = leer_matriz_cuadrada()

print("Traza(A)")
print(traza(A))
```

```python
def e_matriz():
    r = int(input("Número de renglones: "))
    c = int(input("Número de columnas: "))
    print(f"Introduce los {r} renglones (cada uno con {c} números):")
    matriz = []
    for i in range(r):
        fila = input(f"Fila {i+1}: ").split()
        fila = [float(x) for x in fila]
        if len(fila) != c:
            raise ValueError("Número incorrecto de columnas.")
        matriz.append(fila)
    return np.array(matriz)
def transpuesta(A):
    r = A.shape[0]  # número de filas
    c = A.shape[1]  # número de columnas
    T = []  # aquí construiremos la transpuesta
    for j in range(c):
        nueva_fila = []
        for i in range(r):
            nueva_fila.append(A[i][j])
        T.append(nueva_fila)
    return np.array(T)
#Programa principal
print("Matriz A")
A = e_matriz()
print("Transpuesta de A")
print(transpuesta(A))
```

```python
def max_min_matriz(A):
    maximo = np.max(A)
    minimo = np.min(A)
    return maximo, minimo
print("Matriz A")
A = e_matriz()
maximo, minimo = max_min_matriz(A)

print("Máximo elemento de la matriz:", maximo)
print("Mínimo elemento de la matriz:", minimo)
```

```python
import numpy as np

def generar_matriz_cuadrada():
    while True:
        n = int(input("Número de filas (3 a 5): "))
        m = int(input("Número de columnas (3 a 5): "))
        if 3 <= n <= 5 and 3 <= m <= 5:
            if n == m:
                break
            else:
                print("Error: La matriz no es cuadrada. Intenta de nuevo.")
        else:
            print("Dimensiones fuera de rango. Intenta de nuevo.")
    # matriz con valores aleatorios
    t = np.random.randint(-20, 20, (n, m))  # aleatorios entre -20 y 20
    return t
# PROGRAMA PRINCIPAL
t = generar_matriz_cuadrada()
n = t.shape[0]

print("\nMatriz generada:")
print(t)

# 1. Primera fila inversa
print("\nPrimera fila inversa:")
print(t[0][::-1])

# 2. Primera columna inversa
print("\nPrimera columna inversa:")
print(t[:, 0][::-1])

# 3. Primer y último elemento
print("\nPrimer elemento:", t[0, 0])
print("Último elemento:", t[n-1, n-1])

# 4. Máximo, mínimo, promedio diagonal, traza
diag = np.diag(t)
print("\nMáximo:", np.max(t))
print("Mínimo:", np.min(t))
print("Promedio de la diagonal:", np.mean(diag))
print("Traza:", np.trace(t))

# 5. Suma de pares recorriendo filas
suma_pares = 0
for fila in t:
    for x in fila:
        if x % 2 == 0:
            suma_pares += x

print("\nSuma de los elementos pares recorriendo filas:", suma_pares)

# 6. Suma de impares recorriendo columnas
suma_impares = 0
for col in t:
    for x in col:
        if x % 2 != 0:
            suma_impares += x

print("Suma de los elementos impares recorriendo columnas:", suma_impares)

# 7. Promedios de filas y columnas
print("\nPromedio de cada fila:")
print(np.mean(t, axis=1))

print("\nPromedio de cada columna:")
print(np.mean(t, axis=0))

# 8. Antidiagonal y su promedio
antidiagonal = np.fliplr(t).diagonal()
print("\nAntidiagonal:", antidiagonal)
print("Promedio de antidiagonal:", np.mean(antidiagonal))
```

```python
import numpy as np
import matplotlib.pyplot as plt
# Rango de x
x = np.linspace(-3, 3, 400)

# Escogemos constantes representativas
C_pos = [1, 3, 5]      # C > 0
C_zero = [0]           # C = 0
C_neg = [-1, -3, -5]   # C < 0
plt.figure(figsize=(10, 6))
#CURVAS C > 0 (color: azul)
for C in C_pos:
    y = (x - 1) + C * np.exp(-x)
    plt.plot(x, y, label=f"C > 0 (C={C})", color="blue")
#CURVAS C = 0 (color: negro)
for C in C_zero:
    y = (x - 1)
    plt.plot(x, y, label="C = 0", color="black", linewidth=3)
#CURVAS C < 0 (color: rojo)
for C in C_neg:
    y = (x - 1) + C * np.exp(-x)
    plt.plot(x, y, label=f"C < 0 (C={C})", color="red")
# Configuración de la grafica
plt.title("Familia de curvas de la solución: y = (x − 1) + Ce^{-x}")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)
plt.legend()
plt.axhline(0, color="gray")
plt.axvline(0, color="gray")

plt.show()
```
