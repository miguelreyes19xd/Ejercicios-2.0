```python
edad=int(input("ingresa tu edad:"))
print(edad)
if 0<=edad<=11:
  print("eres un niño")
elif 11<edad<=17:
  print("eres un adolescente")
elif 17<edad<=29:
  print("eres un joven")
elif 29<edad<=64:
  print("eres un adulto")
elif edad>=65:
  print("eres un adulto mayor")
elif edad<0:
  print("reinicia el ejercicio")

```
