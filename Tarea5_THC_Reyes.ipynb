Ejercicio 1.- Ingreso a una boveda.

```python
def acceso_boveda(edad_juan, edad_pedro, num_nietos):
  condicion_1 = edad_juan > 35 or edad_pedro < 50
  condicion_2 = (edad_juan + edad_pedro) % 2 == 0
  condicion_3 = num_nietos >= 1
  if condicion_1 and condicion_2 and condicion_3:
    return "SI"
  else:
    return "NO"
juan = int(input("Ingrese la edad de Juan: "))
pedro = int(input("Ingrese la edad de Pedro: "))
nietos = int(input("Ingrese el número de nietos: "))
resultado = acceso_boveda(juan, pedro, nietos)
print("¿Tienen acceso a la boveda?", resultado)
```

Ejercicio 2.- Reloj Mundial.

```python
#En una lista enumeramos las ciudades que usaremos
ciudades = [
  "Tokio, Japón",
  "Pekin, China",
  "Dusambé, Tayikistán",
  "Perth, Australia",
  "Katmandú, Nepal",
  "Vilna, Lituania",
  "Freetown, Sierra Leona",
  "Río de Janeiro, Brasil",
  "Tegucigalpa, Guatemala",
  "La Paz, Baja California, Mexico"]
#ahora en un diccionario guardamos las diferencias en horas
#de cada ciudad
diferencia_horario = {
  "Tokio, Japón": 15,
  "Pekin, China": 14,
  "Dusambé, Tayikistán": 11,
  "Perth, Australia": 14,
  "Katmandú, Nepal": 11.75,
  "Vilna, Lituania": 8,
  "Freetown, Sierra Leona": 6,
  "Río de Janeiro, Brasil": 3,
  "Tegucigalpa, Guatemala": 0,
  "La Paz, Baja California, Mexico": -1}
#Pediremos al usuario una hora en CDMX
hora_cdmx = input("Ingrese la hora actual en CDMX: ")
hm = hora_cdmx.split(":") #aqui dividimos la hora en hora y minuto
hora = int(hm[0])
minuto = int(hm[1])
minutos_mex = hora*60+minuto #convertimos a minutos totales
#el lambda es para para sumar la diferencia horaria
calcular_hora = lambda base, diff: (base+diff*60)% (24*60)
print("hora en difererntes ciudades del mundo")
#en el ciclo for recorremos la lista para calcular la hora
for ciudad in ciudades:
  hora_mex = calcular_hora(minutos_mex, diferencia_horario[ciudad])
#Aqui convertimos los minutos totales en horas y minutos
  h_local = int(hora_mex // 60)
  m_local = int(hora_mex % 60)
  print(f"{ciudad}: {h_local}:{m_local:02d}")
```

5.- MCD y MCM

```python
def mcd_mcm(a,b):
  def mcd(x,y):
    while y != 0:
      x, y = y, x % y
    return x
  def mcm(x,y):
    return (x * y) // mcd(x,y)
  mcd_resultado = mcd(a,b)
  mcm_resultado = abs(a * b) // mcd_resultado
  resultados = {
      "numero 1": a,
      "numero 2": b,
      "mcd": mcd_resultado,
      "mcm": mcm_resultado
  }
  print("Resultados", resultados)
a=int(input("Ingrese el primer numero: "))
b=int(input("Ingrese el segundo numero: "))
mcd_mcm(a,b)
```

```python
datos = [25, "Matematicas", 3.141592, True, 8, False,
         "Nombres", 2.718, "Unam", 0, 9.81, "CdMx",
         7e12, 12.5, "2025", True, "poblacion",
         0.01, 42, False]
def flitrar_datos(datos):
  enteros = []
  cadenas = []
  booleanos = []
  flotantes = []
  for elemento in datos:
    if type(elemento) == int and not isinstance(elemento, bool):
      enteros.append(elemento)
    elif type(elemento) == float:
      flotantes.append(elemento)
    elif type(elemento) == bool:
      booleanos.append(elemento)
    elif type(elemento) == str:
      cadenas.append(elemento)
  print("Enteros:", enteros)
  print("Cadenas:", cadenas)
  print("Booleanos:", booleanos)
  print("Flotantes:", flotantes)
flitrar_datos(datos)
```

7.- formula de Haversine

```python
import math
def haversine(lat1, lon1, lat2, lon2):
  R = 6378.1
  lat1_rad = math.radians(lat1)
  lon1_rad = math.radians(lon1)
  lat2_rad = math.radians(lat2)
  lon2_rad = math.radians(lon2)
  dlat = lat2_rad - lat1_rad
  dlon = lon2_rad - lon1_rad
  hs = (math.sin(dlat/2)**2)
  hs2 = (math.sin(dlon/2)**2)
  a = hs + math.cos(lat1_rad) * math.cos(lat2_rad) * hs2
  distancia = 2 * R * math.asin(math.sqrt(a))
  return distancia
#Ejemplo profe
paris = (48.9, 2.4)
roma = (41.9, 12.5)
print("la distancia es",haversine(*paris, *roma), "km")
#MIs ejmeplos
CDMX = (19.43, -99.13)
Nueva_York = (40.71, -74.01)
Buenos_Aires = (-34.6, -58.41)
tokio = (35.7, 139.7)
print("la distancia es",haversine(*CDMX, *Nueva_York), "km")
print("La distancia es", haversine(*tokio, *Nueva_York), "km")
print("La distancia es", haversine(*CDMX, *Buenos_Aires), "km")
```

8.- La serie de fibinacci

```python
n = int(input("Ingrese un numero: "))
def fibonacci(n):
  if n <= 0:
    print("Error en el numero, vuelve a intentarlo")
    return
  if n == 1:
    return [0]
  elif n == 2:
    return [0,1]
  else:
    serie = [0,1]
    for i in range(2, n):
      serie.append(serie[i-1] + serie[i-2])
    return serie

serie_fibonacci = fibonacci(n)
if serie_fibonacci is not None:
  print("serie de Fibonacci con", n, "elementos:", *serie_fibonacci)
```

10.- Operaciones Vectoriales

```python
a=[2,-1,4]
b=[4,-2,8]
c=[-5,2,1]
def producto_escalar (u,v):
  return sum(x*y for x,y in zip(u,v))
def producto_vectorial (u,v):
  return [
      u[1]*v[2]-u[2]*v[1],
      u[2]*v[0]-u[0]*v[2],
      u[0]*v[1]-u[1]*v[0]
  ]
def revisar_vectores (u,v):
  if producto_escalar(u,v) == 0:
    return"Los vectores son perpendiculares"
  if producto_vectorial(u,v) == [0,0,0]:
    return"Los vectores son paralelos"
  return"Los vectores no son ni perpendiculares ni paralelos"
def operaciones_vectores(u,v):
  print(f"vectores: {u} y {v}")
  print(f"producto escalar: {producto_escalar(u,v)}")
  print(f"producto vectorial: {producto_vectorial(u,v)}")
  print("RElacion", revisar_vectores(u,v))
operaciones_vectores(a,b)
operaciones_vectores(a,c)
operaciones_vectores(b,c)
```
